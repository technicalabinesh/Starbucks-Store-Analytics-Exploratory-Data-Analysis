# Starbucks Store Analytics - Exploratory Data Analysis

This notebook performs a quick exploratory analysis of the `starbucks.csv` dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_style('whitegrid')
```

```python
df = pd.read_csv('starbucks.csv')
df.head()
```

```python
print(f'Rows: {df.shape[0]}, Columns: {df.shape[1]}')
df.info()
```

```python
# Clean column names
df.columns = (
    df.columns
    .str.strip()
    .str.replace('% DV', 'pct_dv', regex=False)
    .str.replace('(', '', regex=False)
    .str.replace(')', '', regex=False)
    .str.replace('/', '_', regex=False)
    .str.replace(' ', '_', regex=False)
    .str.replace('__', '_', regex=False)
    .str.lower()
)
df.columns
```

```python
# Convert percentage columns to numeric values
pct_cols = [c for c in df.columns if 'pct_dv' in c]
for col in pct_cols:
    df[col] = df[col].astype(str).str.replace('%', '', regex=False)
    df[col] = pd.to_numeric(df[col], errors='coerce')

df[pct_cols].head()
```

```python
# Missing values
df.isnull().sum().sort_values(ascending=False)
```

```python
# Summary statistics
df.describe(include='all').T
```

```python
avg_calories = (
    df.groupby('beverage_category', as_index=False)['calories']
    .mean()
    .sort_values('calories', ascending=False)
)
avg_calories
```

```python
plt.figure(figsize=(12, 5))
sns.countplot(data=df, y='beverage_category', order=df['beverage_category'].value_counts().index, palette='viridis')
plt.title('Number of Drinks by Beverage Category')
plt.xlabel('Count')
plt.ylabel('Beverage Category')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['calories'], bins=30, kde=True, color='teal')
plt.title('Calories Distribution')
plt.xlabel('Calories')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include=['number'])
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
